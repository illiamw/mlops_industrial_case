```python
#Variaveis de Ambiente
try:
    SOURCE_CATALOG = dbutils.widgets.get("source_catalog")
except:
    SOURCE_CATALOG = "test"

try:
    TARGET_CATALOG = dbutils.widgets.get("target_catalog")
except:
    TARGET_CATALOG = "prod"

try:
    SCHEMA = dbutils.widgets.get("schema")
except:
    SCHEMA = "falha"

try:
    MODEL = dbutils.widgets.get("model")
except:
    MODEL = "failure_predictor"

source_name = f"{SOURCE_CATALOG}.{SCHEMA}.{MODEL}"
target_name = f"{TARGET_CATALOG}.{SCHEMA}.{MODEL}"
```

```python
import mlflow
from mlflow import MlflowClient

client = MlflowClient()


# 1. Recuperar versão aprovada em TEST
source_version = client.get_model_version_by_alias(
    source_name,
    "Candidate"
)

# 2. Copiar versão para PROD
new_version = client.copy_model_version(
    src_model_uri=(
        f"models:/{source_name}/{source_version.version}"
    ),
    dst_name=target_name
)

# 3. Atualizar alias Champion em PROD
client.set_registered_model_alias(
    name=target_name,
    alias="Champion",
    version=new_version.version
)

print(f"Modelo publicado: {target_name}")
print(f"Versão: {new_version.version}")
```
